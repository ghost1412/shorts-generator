# ⚡ ShortsFlow AI Studio on Google Colab
Free GPU, no local install. Turn a YouTube link into vertical Shorts with face tracking and animated captions, or generate a short from an idea.

1. **Runtime → Change runtime type → T4 GPU**
2. Add `GEMINI_API_KEY` (and optionally `PEXELS_API_KEY`) in the 🔑 **Secrets** panel on the left
3. Run the cells below and open the `gradio.live` link

⭐ If this helps, star the repo: https://github.com/ghost1412/shorts-generator

```
#@title 2. Load API keys from Colab Secrets
import os
from google.colab import userdata
for key in ["GEMINI_API_KEY", "PEXELS_API_KEY", "HF_API_KEY", "ELEVENLABS_API_KEY", "OPENAI_API_KEY"]:
    try:
        os.environ[key] = userdata.get(key)
        print(f"✅ {key}")
    except Exception:
        print(f"– {key} not set")
```
